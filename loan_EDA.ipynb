```python
## visualization of target data to check ratio
sns.countplot(data  =df, x='loan_status')
plt.title("Loan Approval Distribution")
plt.show()
```

```python
## Separate Numerical & Categorical Columns
cat_cols = df.select_dtypes(include = "object").columns
num_cols = df.select_dtypes(include = "number").columns
print("Numerical Columns:", num_cols)
print("Categorical Columns:", cat_cols)
```

```python
## Distribution of Numerical Features
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()
```

```python
## categorical counts

for col in cat_cols:
    print(df[col].value_counts())
    print("-"*30)
```

```python
## Credit Score vs Loan Status
sns.boxplot(x='loan_status', y='credit_score', data=df)
plt.show()
```

```python
## income vs Loan Status
sns.boxplot(x='loan_status', y='annual_income', data=df)
plt.show()
```

```python
df[cat_cols].columns
```

```python
## Categorical Feature vs Loan Status
cat_cols =['occupation_status', 'product_type', 'loan_intent']
for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, hue='loan_status', data=df)
    plt.title(f"{col} vs Loan Status")
    plt.xticks(rotation=45)
    plt.show()
```

```python
## heatmap corelation for detecting multicollinearity
plt.figure(figsize=(12,8))
sns.heatmap(df[num_cols].corr(), cmap='BrBG',fmt='.2f', annot=True)
plt.show()
```
